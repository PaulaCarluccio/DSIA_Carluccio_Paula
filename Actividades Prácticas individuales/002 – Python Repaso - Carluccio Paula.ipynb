### Ejercicio 3 – Listas y bucles

```
lista = [1, 2, 3, 4, 5]

for i in lista:
    print(i)

suma = sum(lista)
promedio = suma / len(lista)

print(f"La suma total de los números es: {suma} y el promedio es: {promedio}")
```
